```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, mean_squared_error
from sklearn.naive_bayes import GaussianNB
```

```python
iris = load_iris()
display(iris)
```

```python
X = iris.data
y = iris.target
display(X)
display(y)
```

```python
classes = np.unique(y)
print(classes)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, random_state = 42)
print(f"Train Size: \nX = {X_train.shape}\tY = {y_train.shape}")
print(f"Test size = \nX={X_test.shape}\tY = {y_test.shape}")
```

```python
X_c = {}
priors = {}
means = {}
var = {}
for i in classes:
    X_c[i] = X_train[y_train == i]
    priors[i] = X_c[i].shape[0]/X_train.shape[0]
    means[i] = np.mean(X_c[i], axis = 0)
    var[i] = np.var(X_c[i], axis = 0)
print(f"Priors = {priors}")
print(f"means = {means}")
print(f"var = {var}")
```

```python
predictions = []
for x in X_test:
    scores = {}
    for i in classes:
        log_prior = np.log(priors[i])
        eps = 1e-9 
        e = np.exp(-((x - means[i]) ** 2) / ( 2 * var[i] + eps))
        pdf = e / np.sqrt(2 * np.pi * var[i] + eps)
        log_liklihood = np.sum( np.log(pdf + 1e-9))
        scores[i] = log_prior + log_liklihood
    predictions.append(max(scores, key = scores.get))
print(predictions)
```

```python
# 1. Classification Error Rates
accuracy = accuracy_score(y_test, predictions)
classification_error = 1.0 - accuracy

print("=== Basic Metrics ===")
print(f"Model Accuracy     : {accuracy * 100:.2f}%")
print(f"Classification Error Rate: {classification_error * 100:.2f}%")

# 2. Root Mean Squared Error (RMSE)
# Note: RMSE computes the numerical distance mismatch between the actual class indices and the predicted class numbers.
mse = mean_squared_error(y_test, predictions)
rmse = np.sqrt(mse)
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")

# 3. Detailed Precision, Recall, and F1-Scores
print("\n=== Classification Report ===")
print(classification_report(y_test, predictions, target_names=iris.target_names))

# 4. Confusion Matrix Plotting
cm = confusion_matrix(y_test, predictions)
plt.figure(figsize=(6, 4.5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.title('Naive Bayes Confusion Matrix')
plt.ylabel('Actual Label')
plt.xlabel('Predicted Label')
plt.show()
```
